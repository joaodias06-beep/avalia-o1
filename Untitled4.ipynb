```python
senha_correta = "1234"
tentativas = 0

while tentativas < 3:
    senha = input("Digite sua senha: ")
    tentativas += 1

    if senha == senha_correta:
        print("Bem-vindo! Acesso autorizado.")
        break
    elif tentativas < 3:
        print(f"Senha incorreta. Você ainda tem {3 - tentativas} tentativa(s).")
    else:
        print("Acesso bloqueado. Você excedeu o limite de 3 tentativas.")
```
